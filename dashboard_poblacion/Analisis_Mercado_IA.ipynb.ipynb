##CREACION DEL PROYECTO

```python
def promedio(columna):
    """Calcula el promedio de una columna."""
    return np.mean(columna)


def filtrar(tabla, columna, opcion):
    """Filtra los datos según la opción seleccionada."""
    if opcion != "Todos":
        tabla = tabla[tabla[columna] == opcion]
    return tabla
```

##RESUMEN GENERAL DEL MERCADO

```python
salario = datos["salary"]
experiencia = datos["years_experience"]

print("Ofertas laborales:", len(datos))
print("Salario promedio:", round(promedio(salario), 2))
print("Años de experiencia promedio:", round(promedio(experiencia), 1))
```

##CALCULOS CON NUMPY

```python
estadisticas = {
    "Media": round(np.mean(salario), 2),
    "Mediana": round(np.median(salario), 2),
    "Mínimo": np.min(salario),
    "Máximo": np.max(salario),
    "Desviación estándar": round(np.std(salario), 2)
}

for nombre in estadisticas:
    print(nombre, ":", estadisticas[nombre])
```

##SALARIO PROMEDIO POR NIVEL DE EXPERIENCIA

```python
niveles = ["Entry", "Mid", "Senior"]
salarios = {}

for nivel in niveles:
    grupo = datos[datos["experience_level"] == nivel]

    if len(grupo) > 0:
        salarios[nivel] = promedio(grupo["salary"])

for nivel in salarios:
    print(nivel, ":", round(salarios[nivel], 2))
```

##GRAFICO

```python
pd.Series(salarios).plot(kind="bar")

plt.title("Salario promedio por experiencia")
plt.xlabel("Nivel de experiencia")
plt.ylabel("Salario promedio")
plt.show()
```

##ANALISIS DE OFERTAS Y SALARIOS POR AÑO

```python
años = sorted(datos["job_posting_year"].unique())

ofertas_anuales = {}
salarios_anuales = {}

for año in años:
    grupo = datos[datos["job_posting_year"] == año]

    ofertas_anuales[año] = len(grupo)
    salarios_anuales[año] = promedio(grupo["salary"])

print("Ofertas por año:")
print(ofertas_anuales)

print("Salarios promedio por año:")
print(salarios_anuales)
```

##GRAFICO DE LINEAS

```python
pd.Series(ofertas_anuales).plot(kind="line")

plt.title("Ofertas laborales por año")
plt.xlabel("Año")
plt.ylabel("Cantidad de ofertas")
plt.show()
```

##GRAFICO DEL AREA

```python
pd.Series(salarios_anuales).plot(kind="area")

plt.title("Evolución del salario promedio")
plt.xlabel("Año")
plt.ylabel("Salario promedio")
plt.show()
```

##CARGOS PROFESIONALES MAS FRECUENTES

##ANALISIS DE HABILIDADES TECNOLOGICAS

```python
habilidades = [
    "skills_python",
    "skills_sql",
    "skills_ml",
    "skills_deep_learning",
    "skills_cloud"
]

for habilidad in habilidades:
    cantidad = datos[datos[habilidad] == 1]

    porcentaje = len(cantidad) / len(datos) * 100

    print(habilidad, ":", round(porcentaje, 2), "%")
```
